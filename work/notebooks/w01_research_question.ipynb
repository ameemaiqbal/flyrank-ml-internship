A quick look at the starter dataset: comparing declining vs. stable/growing pages on average position and position spread, as an early signal that ranking instability may be worth investigating further with the full time-series data.

```python
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/ameemaiqbal/flyrank-ml-internship"
REPO_DIR = "flyrank-ml-internship"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)

print("Working dir:", os.getcwd())
assert os.path.exists("data/raw/content_refresh_anonymized.csv"), "starter CSV not found — check the repo cloned correctly"
print("Ready.")
```

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

declining = df[df["trend_direction"] == "down"]
stable = df[df["trend_direction"].isin(["stable", "up"])]

print(f"Total pages: {len(df):,}")
print(f"Declining pages: {len(declining):,} ({len(declining)/len(df):.1%})")
print(f"Avg position — declining pages: {declining['avg_position'].mean():.1f}")
print(f"Avg position — stable/growing pages: {stable['avg_position'].mean():.1f}")
print(f"Position spread (std dev) — declining: {declining['avg_position'].std():.1f}")
print(f"Position spread (std dev) — stable/growing: {stable['avg_position'].std():.1f}")
```

Note: these numbers use avg_position as a single snapshot value, not a day-to-day time series, so they don't directly test ranking volatility. Interestingly, declining pages in this sample actually show lower average position and lower spread than stable/growing pages, the opposite of what my hypothesis predicts. This is likely because this CSV lacks the daily granularity needed to compute real volatility (unlike the full warehouse data used in Week 3). This discrepancy is itself a reason the lane is worth pursuing: the real test requires the full daily dataset, not this snapshot.
